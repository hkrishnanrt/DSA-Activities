1. Grocery Store Billing
A grocery store wants to automate billing.
● Take the names and prices of items from the user until they type "done".
● Print the total bill.
● If total bill > ₹500, apply 10% discount

```python
bill=0
i=1
while i>0:
  name=input("Enter the item : ").lower()
  if name!='done':
    price=float(input("Enter the price : "))
    print()
    bill+=price
  else:
    break
if bill>500:
  total_price=bill-(0.1*bill)
  print(f"Total bill = {total_price}")
else:
  print(f"Total bill = {bill}")

```

2. Exam Result Calculator
A school stores student marks in a list.
● Input marks for 5 subjects.
● Calculate percentage.
● Print grade:
○ ≥ 90: A
○ ≥ 75: B
○ ≥ 50: C
○ else: Fail


```python
marklist=[]
total_mark=0
for i in range(5):
  mark=int(input("Enter the mark "))
  marklist.append(mark)
  total_mark+=mark
Percentage=(total_mark/500)*100
print()
print(f"Percentage : {Percentage:.2f}")
if Percentage>=90:
  print("Grade : A")
elif Percentage>=75:
  print("Grade : B")
elif Percentage >=50:
  print("Grade : C")
else:
  print("Fail")
```

4. Cricket Score Tracker
Store player scores in a dictionary.
● Add 5 players with scores.
● Find and print top scorer’s name and score.
● Update a player’s score after a match.


```python
score_tracker={}

for i in range(5):
  name=input("Enter the player name : ")
  score=int(input("Enter the score : "))
  score_tracker[name]=score
print(score_tracker)
print()

for k,v in score_tracker.items():
  top_score=0
  if v>top_score:
    top_score=v
    name=k
print(f"Top Scorer : {name} Runs : {top_score}")
print()
player_update=input("Enter the player to update score : ")
if player_update in score_tracker:
  new_score=int(input(f"Enter {player_update}'s new score "))
  score_tracker[player_update]=new_score
print("Updated Scores : ",score_tracker)

```

5. Restaurant Order System
● Menu stored in dictionary (item: price).
● Take multiple orders from user.
● Print final bill amount.
● If order contains "Pizza", give ₹50 off

```python
menu={'Burger':200,'Pizza':120,'Sandwich':100,'Coffee':40,'Tea':20}
print("Menu")
for k,v in menu.items():
  print(f"{k} : {v}")
print()

order_total=0
orders={}
while True:
  order_item=input("Emter the item you want to order ").capitalize()
  if order_item=='Done':
    break

  if order_item in menu:
    quantity=int(input("Enter the quantity "))
    cost=menu[order_item] * quantity
    order_total+=cost
    print()
  if order_item in orders:
    orders[order_item]+=quantity
  else:
    orders[order_item]=quantity
print()

if 'Pizza' in orders:
  order_total=order_total-50

for k,v in orders.items():
  print (f"{k} * {v} : {menu[k] * v}")
print(f"Total bill : {order_total}")
```

7. Online Shopping Cart
● Create a cart as a list.
● Allow user to add items, remove items, and view cart.
● Exit when user types "checkout"

```python
cart=[]
print("Options : ADD,REMOVE,VIEW,CHECKOUT")
print()

while True:
  option=input("Enter the option ").lower()
  print()

  if option=='add':
    item=input("Enter the item to be added ")
    cart.append(item)
    print(f"{item} is added to the cart")
    print()

  elif option=='remove':
    item=input("Enter the item to be removed ")
    if item in cart:
      cart.remove(item)
      print(f"{item} is removed from the cart")
    else:
      print(f"{item} is not found in the cart")
      print()

  elif option=='view':
    print(f"Cart : {cart}")

  elif option=='checkout':
    print("Checkout completed")
    print(f"Cart : {cart}")
    print()
    break

  else:
    print("Invalid option")
```

9. Cinema Ticket Booking System
Create a program where users can book movie tickets. A theater has 50 seats in total. Each
ticket costs ₹120.
● Ask the user how many tickets they want.
● If enough seats are available → confirm booking and reduce available seats.
● If not enough seats → show an error message.
● Show remaining seats after each booking.

```python
seats=50
cost=120
while seats>0:

  tickets=int(input("Enter the number of tickets "))

  if tickets<=seats:
    total_cost=cost*tickets
    seats=seats-tickets
    print("Booking Confirmed")
    print(f"Total Cost : {total_cost}")
    print(f"Remaining Seats : {seats}\n")

  else:
    print(f"Only {seats} available, Cannot book {tickets} tickets")

print("Housefull")
```
