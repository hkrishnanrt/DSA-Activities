1. Dataset: Monthly temperatures in a city (in °C): 12, 15, 14, 16, 18, 20, 22, 19,
17, 15, 13, 10 Questions:
a) Calculate the mean temperature.
b) Calculate the median temperature.
c) Find the mode of the temperature data.
d) Range and Variance

```python
temperature=[12,15,14,16,18,20,22,19,17,15,13,10]

mean = sum(temperature) / len(temperature)

temperature_sorted = sorted(temperature)
n = len(temperature_sorted)
if n % 2 == 0:
    median = (temperature_sorted[n//2 - 1] + temperature_sorted[n//2]) / 2
else:
    median = temperature_sorted[n//2]

mode = None
max_count = 0
for value in temperature_sorted:
    count = temperature_sorted.count(value)
    if count > max_count:
        max_count = count
        mode = value

data_range = max(temperature) - min(temperature)
variance = sum((x - mean) ** 2 for x in temperature) / len(temperature)

print("The Mean is:", round(mean, 2))
print("The Median is:", median)
print("The Mode is :", mode)
print("The Range is:", data_range)
print("Variance is :", round(variance, 2))
```

2. Dataset: Heights of students (in cm): 150, 160, 165, 170, 155, 160, 162, 168
Questions: a) Calculate the range of the heights.
b) Calculate the variance of the heights.
c) Standard Deviation


```python
Height=[150,160,165,170,155,160,162,168]

Range=max(Height)-min(Height)

mean = sum(Height) / len(Height)
variance = sum((x - mean) ** 2 for x in Height) / len(Height)
std=(variance**0.5)

print("Range of the heights : ",Range)
print(f"Variance of the heights : {variance:.3f}")
print(f"Standard deviation : {std:.3f}")
```

4. Dataset: Advertising Spending (in $1000s): 10, 20, 15, 25, 30; Sales (in
$1000s): 100, 150, 120, 180, 200
Questions: a) Calculate the covariance between advertising spending and sales.
b) Correlation

```python
spend = [10, 20, 15, 25, 30]
sales = [100, 150, 120, 180, 200]

n = len(spend)

mean_spend = sum(spend) / n
mean_sales = sum(sales) / n

cov_xy = sum((spend[i] - mean_spend) * (sales[i] - mean_sales) for i in range(n)) / n

var_x = sum((spend[i] - mean_spend) ** 2 for i in range(n)) / n
var_y = sum((sales[i] - mean_sales) ** 2 for i in range(n)) / n

corr_xy = cov_xy / ((var_x ** 0.5) * (var_y ** 0.5))


print("Covariance:", cov_xy)
print("Correlation:", corr_xy)
```

5. Dataset: Study Hours: 2, 4, 6, 8, 10; Exam Scores: 55, 60, 65, 70, 75
Questions: a) Calculate the correlation coefficient between study hours and exam
scores.
b) Interpret the result.
c) Quartiles and IQR


```python
def percentile(data, percent):
    data = sorted(data)
    k = (len(data) - 1) * (percent / 100)
    f = int(k)
    c = f + 1
    if c >= len(data):
        return data[f]
    return data[f] + (data[c] - data[f]) * (k - f)

def quartiles(data):
    Q1 = percentile(data, 25)
    Q2 = percentile(data, 50)
    Q3 = percentile(data, 75)
    IQR = Q3 - Q1
    return Q1, Q2, Q3, IQR

Study_hours = [2, 4, 6, 8, 10]
Exam_scores = [55, 60, 65, 70, 75]
n = len(Study_hours)

mean_hours = sum(Study_hours) / n
mean_scores = sum(Exam_scores) / n

cov_xy = sum((Study_hours[i] - mean_hours) * (Exam_scores[i] - mean_scores) for i in range(n)) / (n-1)
var_x = sum((Study_hours[i] - mean_hours) ** 2 for i in range(n)) / (n-1)
var_y = sum((Exam_scores[i] - mean_scores) ** 2 for i in range(n)) / (n-1)

corr_xy = cov_xy / ((var_x ** 0.5) * (var_y ** 0.5))

print("Correlation:", corr_xy)
print("Study Hours Quartiles & IQR:", quartiles(Study_hours))
print("Exam Scores Quartiles & IQR:", quartiles(Exam_scores))




```

6. Dataset: Monthly savings (in $): 200, 300, 400, 500, 600, 700, 800, 900, 1000,
1100 Questions: a) Calculate the first quartile (Q1) and third quartile (Q3).
b) Calculate the interquartile range (IQR).
c) Detecting Outliers


```python
# Function to compute median manually
def median(data):
    data = sorted(data)
    n = len(data)
    mid = n // 2
    if n % 2 == 0:  # even length
        return (data[mid - 1] + data[mid]) / 2
    else:  # odd length
        return data[mid]

# Function to compute quartiles and IQR
def quartiles(data):
    data = sorted(data)
    n = len(data)

    Q2 = median(data)

    # Split into lower and upper halves
    if n % 2 == 0:
        lower = data[:n//2]
        upper = data[n//2:]
    else:
        lower = data[:n//2]
        upper = data[n//2+1:]

    Q1 = median(lower)
    Q3 = median(upper)
    IQR = Q3 - Q1

    return Q1, Q2, Q3, IQR

def detect_outliers(data):
    Q1, Q2, Q3, IQR = quartiles(data)
    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR
    outliers = [x for x in data if x < lower_fence or x > upper_fence]
    return outliers, lower_fence, upper_fence

savings = [200, 300, 400, 500, 600, 700, 800, 900, 1000, 1100]

Q1, Q2, Q3, IQR = quartiles(savings)
outliers, lower_fence, upper_fence = detect_outliers(savings)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
print("Outliers:", outliers if outliers else "No outliers")
```

9. Dataset: Sample weights (in kg) of a new diet program: 70, 68, 72, 71, 69, 73,
72, 74, 70, 71 Questions: a) Test the hypothesis that the mean weight after the
diet program is equal to 70 kg at a 5% significance level. b) Z-Scores


```python
Weight=[70,68,72,71,69,73,72,74,70,71]

mean = sum(Weight) / len(Weight)
n = len(Weight)
variance = sum((Weight[i] - mean)**2 for i in range(n)) / n-1
std = variance**0.5

mu0=70
t_stat=(mean-mu0)/(std/(n**0.5))
t_critical=2.262
if abs(t_stat)>t_critical:
  print("Reject null hypothesis(Mean is significantly different from 70)")
else:
  print("Failed to reject null hypothesis (Mean is not significantly different from 70)")

a = [(Weight[i] - mean) / std for i in range(n)]

print(f"Z-score : {a} ")
```
